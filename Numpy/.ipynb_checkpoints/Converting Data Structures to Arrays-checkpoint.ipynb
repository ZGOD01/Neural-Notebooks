```python
import numpy as np
```

### 1. Converting Data Structures to Numpy Array

#### 1.1 Converting a Tuple to an array

```python
tpl =(1,2,3 )
print(tpl)
print(type(tpl))

print("-"*30)

arr = np.array(tpl)
print(arr)
print(type(arr))
```

#### 1.2 Converting a List to an array

```python
lst =[1,2,3 ]
print(lst)
print(type(lst))

print("-"*30)

arr = np.array(lst)
print(arr)
print(type(arr))
```

#### 1.3 Converting a List to an array

dct = {1 : 'apple' , 2 : 'orange' , 3 : 'cherry'}

### 2. Converting the Datatypes of Numpy Array

#### 2.1 Create a Array of integers and then converting into other Datatypes 

```python
arr = np.array([1,2,3,4,5])
print(arr)

print('-'*10)

arr_ = arr.astype('float')
print(arr_)

print('-'*10)

arr_ = arr.astype('str')
print(arr_)

print('-'*10)

arr_ = arr.astype('bool')
print(arr_)
```

#### 2.1 Create a Array of Float and then converting into other Datatypes 

```python
arr = np.array([1.,2.,3.,4.,5.])
print(arr)

print('-'*10)

arr_ = arr.astype('int')
print(arr_)

print('-'*10)

arr_ = arr.astype('str')
print(arr_)

print('-'*10)

arr_ = arr.astype('bool')
print(arr_)
```

#### 2.1 Create a Array of String and then converting into other Datatypes 

```python
arr = np.array(['1','2','3','4','5']) # Special Case - if the arr is like ['1','A','3','C','5']  it will not be converted other datatypes
print(arr)

print('-'*10)

arr_ = arr.astype('int')
print(arr_)

print('-'*10)

arr_ = arr.astype('float')
print(arr_)

print('-'*10)

arr_ = arr.astype('bool')
print(arr_)
```

#### 2.1 Create a Array of Boolean and then converting into other Datatypes 

```python
arr = np.array([True , False , False , False , True])
print(arr)

print('-'*10)

arr_ = arr.astype('int')
print(arr_)

print('-'*10)

arr_ = arr.astype('float')
print(arr_)

print('-'*10)

arr_ = arr.astype('str')
print(arr_)
```
